***Problem Statement***

The dataset containg missing value ,inconsistent data type and unnecessary infoemation that can affect analysis and machine learning performance.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("TitanicDataset.csv")
print("Dataset loaded successfully.")

print("\nInitial DataFrame head:")
display(df.head())
print("\nInitial DataFrame info:")
display(df.info())
```

```python
print("DataFrame shape:")
display(df.shape)
```

```python
print("DataFrame information:")
display(df.info())
```

```python
print("Missing values before handling:")
display(df.isnull().sum())

# Fill missing 'Age' values with the median
if 'Age' in df.columns and df['Age'].isnull().any():
    median_age = df['Age'].median()
    df['Age'] = df['Age'].fillna(median_age)
    print(f"Filled missing 'Age' values with median: {median_age}")

# Fill missing 'Embarked' values with the mode
if 'Embarked' in df.columns and df['Embarked'].isnull().any():
    mode_embarked = df['Embarked'].mode()[0]
    df['Embarked'] = df['Embarked'].fillna(mode_embarked)
    print(f"Filled missing 'Embarked' values with mode: {mode_embarked}")

# Fill missing 'Fare' values with the median
if 'Fare' in df.columns and df['Fare'].isnull().any():
    median_fare = df['Fare'].median()
    df['Fare'] = df['Fare'].fillna(median_fare)
    print(f"Filled missing 'Fare' values with median: {median_fare}")

print("\nMissing values after handling (Age, Embarked, Fare):")
display(df.isnull().sum())
```

```python
# Drop 'Cabin' column due to a high number of missing values (or if deemed unnecessary)
if 'Cabin' in df.columns:
    # Determine if 'Cabin' has a significant number of missing values to warrant dropping
    # For this example, let's assume we drop it if it has more than 50% missing data
    if df['Cabin'].isnull().sum() / len(df) > 0.5:
        df = df.drop("Cabin", axis=1)
        print("Dropped 'Cabin' column due to many missing values.")
    else:
        print("'Cabin' column retained (less than 50% missing values).")
else:
    print("'Cabin' column not found in DataFrame.")
```

```python
# Check for and remove duplicate rows
duplicate_rows = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_rows}")

if duplicate_rows > 0:
    df = df.drop_duplicates()
    print(f"Removed {duplicate_rows} duplicate rows. New shape: {df.shape}")
else:
    print("No duplicate rows found.")
```

```python
# Convert 'Age' to numeric, coercing errors to NaN if any non-numeric values exist
if 'Age' in df.columns:
    df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
    print("Converted 'Age' to numeric type.")

# Convert 'Fare' to numeric, coercing errors to NaN if any non-numeric values exist
if 'Fare' in df.columns:
    df['Fare'] = pd.to_numeric(df['Fare'], errors='coerce')
    print("Converted 'Fare' to numeric type.")

# Convert 'Sex' to category type
if 'Sex' in df.columns:
    df['Sex'] = df['Sex'].astype('category')
    print("Converted 'Sex' to category type.")

# Convert 'Embarked' to category type
if 'Embarked' in df.columns:
    df['Embarked'] = df['Embarked'].astype('category')
    print("Converted 'Embarked' to category type.")

# Also convert 'Survived' and 'Pclass' to category as per original plan
if 'Survived' in df.columns:
    df['Survived'] = df['Survived'].astype('category')
    print("Converted 'Survived' to category type.")

if 'Pclass' in df.columns:
    df['Pclass'] = df['Pclass'].astype('category')
    print("Converted 'Pclass' to category type.")

print("\nDataFrame info after type conversions:")
display(df.info())
```

```python
# Rename columns for consistency and readability
new_column_names = {
    "PassengerId": "passenger_id",
    "Survived": "survived",
    "Pclass": "passenger_class",
    "Name": "name",
    "Sex": "sex",
    "Age": "age",
    "SibSp": "siblings_spouses",
    "Parch": "parents_children",
    "Ticket": "ticket",
    "Fare": "fare",
    "Embarked": "embarked"
}

# Apply renaming only to columns that exist in the DataFrame
df = df.rename(columns={col: new_name for col, new_name in new_column_names.items() if col in df.columns})

print("Columns renamed.")
print("\nDataFrame columns after renaming:")
display(df.columns)
```

```python
print("Final check for missing values:")
display(df.isnull().sum())
```

```python
print("Cleaned DataFrame head:")
display(df.head())
```

```python
print("Final DataFrame info:")
display(df.info())
```

```python
output_filename = "Titanic_Cleaned.csv"
df.to_csv(output_filename, index=False)
print(f"Cleaned dataset saved as {output_filename}")
```
